### Chapter 5 - Formulas and Concepts

#### Chapter 5.1

$$ q_{substance\ a} + q_{substance\ b} = 0 $$
$$ \text{specific heat}\  c = \frac{q}{m\Delta T} $$
$$ \Delta T = T_{final} - T_{initial} = T_f - T_i $$
$$
\begin{align}
q &= c \cdot m \cdot \Delta T \\ 
&= c \cdot m \cdot (T_{final} - T_{initial}) \\
&= m c \Delta T \quad \text{Can be pronounced m-cat}
\end{align}
$$
$$ (c_a \cdot m_a \cdot \Delta T_a) + (c_b \cdot m_b \cdot \Delta T_b) = 0 $$
$$ (c_a \cdot m_a \cdot (T_{a\ final} - T_{a\ initial})) + (c_b \cdot m_b \cdot (T_{b\ final} - T_{b\ initial})) = 0 $$
$$ (c_a \cdot m_a \cdot (T_{a\ final} - T_{a\ initial})) = - (c_b \cdot m_b \cdot (T_{b\ final} - T_{b\ initial})) $$

#### Chapter 5.3

$$ \begin{align}
q &:= \text{heat (heat flow into system is positive)} \\
q_p &:= \text{heat under constant P} \\
w &:= \text{work (work done on system is positive, by system is negative)} \\
U &:= \text{internal energy} (\equiv E) \\
P &:= \text{pressure} \\
V &:= \text{volume} \\
H &:= \text{enthalpy} \\
\\
\Delta U &= q + w \\
H &= U + PV \text{ (unmeasurable)} \\
\Delta H &= \Delta U + \Delta (PV) \text{ (measurable, negative for exothermic)} \\
&= q + w + \Delta (PV) \\
P \Delta V &= -w \\
\Delta H &= \Delta U + P \Delta V \text{ (assuming constant P)} \\
&= q_p + w + - w \\
&= q_p \\
\\
1\text{ bar} &= 0.987\text{ atm} \\
\Delta H\degree &:= \text{Enthalpy change under standard conditions/standard state.} \\
&\qquad\text{Book uses 1 atm and 298.15 K (25 $\degree$C), slightly different than IUPAC.} \\
&\qquad\text{For solutions the standard conditions are 1 M molarity instead of 1 atm.} \\
\Delta H\rlap\degree_c &:= \text{Standard enthalpy of combustion of 1 mol under standard conditions} \\
\Delta H\rlap\degree_f &:= \text{Standard enthalpy of formation of 1 mol under standard conditions} \\
&\qquad\text{This is formation of 1 mol of a substance from free elements.} \\
\end{align} $$

**Hess’s law:** If a process can be written as the sum of several stepwise processes, the enthalpy change of the total process equals the sum of the enthalpy changes of the various steps.

If $n$ are the stoichiometric coefficients, then...
$$ \begin{align}
\Delta H \rlap \degree _{reaction} &=
        \sum{n \cdot \Delta H \rlap \degree _f (\text{products})}
        - \sum{n \cdot \Delta H \rlap \degree _f (\text{reactants})}
\end{align} $$

First, let's determine the heat capacity $C_{sys}$ of the system as a whole.
$$ \begin{align}

If the 3.21 g of NH<sub>4</sub>NO<sub>3</sub> in Example 5.6 were dissolved in 100.0 g of water under the same conditions, how much would the temperature change? Explain your answer.  In Example 5.6, it is dissolved in 50.0 g of water instead and the temperature drops from 24.9&deg;C to 20.3&deg;C, a drop of 4.6&deg;C, absorbing 1.0 kJ of energy.

```python
display(24.9 - 20.3)
q_example_5_6 = -((3.21 + 50.0) * 4.184 * (20.3 - 24.9))
q_rounded_5_6 = 1000
q_book_intermediate_5_6 = 4.184 * 53.2 * -4.6
display(f'q from 5.6: {q_example_5_6}, rounded: {q_rounded_5_6}')
dt = q_example_5_6 / ((100.0 + 3.21) * 4.184)
dt_from_rounded = q_rounded_5_6 / ((100.0 + 3.21) * 4.184)
dt_from_intermediate = q_book_intermediate_5_6 / ((100.0 + 3.21) * 4.184)
display(f'dT: {dt}, from round: {dt_from_rounded}, from inter: {dt_from_intermediate}')
display(f'T final: {24.9 - dt}, T final from round: {24.9 - dt_from_rounded}')
display(f'Short T final 1: {24.9 - ((24.9 - 20.3) / 2)}')
display(f'Short T final 2: {24.9 - (1.0 * 1000 / ((100.0 + 3.21) * 4.184))}')
display(f'Short q: {-(50.0 + 3.21) * 4.184 * (20.3 - 24.9)}')
display(f'Short T final 3: {24.9 - (1024.1 / ((100.0 + 3.21) * 4.184))}')
```
