**Main group elements** aka **representative elements** - 
Groups 1, 2, and 13 through 18, where last electron added is in an **s** or **p** orbital.  Valance electrons are those in highest shell ***n***.  (This means that **d** electrons added in the same periodic table row are core, not valence, electrons, because their shell is *n - 1*.)  Cations formed by removing last-added electrons.

**Transition elements** aka **transition elements** - 
Typically groups 3 through 12 (though IUPAC makes exceptions), where last electron added is in a **d** orbital.  Valence electrons are those in ***n*s** and **(*n - 1*)d**.  Cations formed by removing ***n*s** electrons first.

**Inner transition elements** - 
Lanthanide and Actinide series, where last electron added is in an **f** orbital, plus Lanthanum and Actinium themselves.  Valence electrons are those in ***n*s**, **(*n - 1*)d**, and **(*n - 2*)f**.  Cations formed by removing ***n*s** electrons first and then [maybe, unclear to me but example 6.11(e) points that way] the **(*n - 1*)d** electrons.


Nice quote from Chapter 6.4:
> It is the loss, gain, or sharing of valence electrons that defines how elements react.

**Atomic radii** trends:
- Increases down a group.
- Decreases across a period (more electron-electron repulsion aka **shielding**.)
- A U shape in the transition elements.

**Ionization Energy** trends:
- Descreases down a group.
- Increases across a period.
- Decrease bumps across a period where you jump to an easier to remove electron.
  - Jump to electron in higher $l$, higher energy subshell.
  - Jump to electron with higher energy because it is paired in the same orbital.
- Increases as you remove more electrons, $IE_n < IE_{n+1}$.
- Increases a lot as you start to remove core, not valence, electrons.

**Electron Affinity** trends (decreases can mean increasingly negative):
- Increases down a group.
- May descrease period 2 to 3 due to small size of $2s$ shell and greater electron repulsion.
- Decreases across a period.
- Increase bumps across a period where you add a higher energy electron.
  - Jump to electron in higher $n$, next shell, e.g. $EA_1$ in group 18.
  - Jump to electron in higher $l$, next subshell, e.g. $EA_1$ in group 2.
  - Jump to a paired electron in the same orbital, e.g. $EA_1$ in group 15.

Negative electron affinity means "gimme that electron".

**Isoelectronic** - Atoms and ions with the same electron configuration.

#### Ex 6.17
What does it mean to say that the energy of the electrons in an atom is quantized?

It means that the energy of an electron in an atom cannot take on just any value along a continuous range of values.  Instead, it must take on discrete values, only those values, and nothing in between.  This is due to 3-D standing waves in the wave equation underlying the electron.

#### Ex 6.29
How are the Bohr model and the Rutherford model of the atom similar? How are they different?

They are similar in that they both envision the atom as having electrons orbiting a tiny nucleus containing most of the mass.  They are different in that the Rutherford model allowed the electrons to take any energy and orbital radius, whereas the Bohr model quantized the orbits to certain discrete energy levels and radii.

#### Ex 6.31
How are the Bohr model and the quantum mechanical model of the hydrogen atom similar? How are they different?

They are similar in that they both envison most of the mass of the atom as being concentrated in a nucleus in the center with electrons surrounding it.  They are also similar in that they restrict the electron energy levels to discrete shells with discrete energy levels.  They differ in that the QM model adds three additional quantum numbers offering discrete energy levels.  They differ in that the QM model can predict behavior when additional electrons are added in ions.  However, the biggest difference is that the Bohr model still assumes electrons are point particles orbiting in planetary like orbits.  The QM model assumes electrons have a wave nature explaining the quantum numbers, and that electron location is not in circular orbits but is governed by a wave function whose square gives the probability of an electron being at any given location, as if they are smeared out into clouds.

#### Ex 6.33
Describe the properties of an electron associated with each of the following four quantum numbers: n, l, ml, and ms.

**n** - Shell.  Overall energy level and rough distance from the nucleus.

**l** - Subshell.  Orbital angular momentum determining the shape of the orbitals and a more exact energy level.

**m<sub>l</sub>** - Orbital.  Magnetic moment or something of the orbital which determines its spatial orientation.

**m<sub>s</sub>** - Electron.  Quantum spin up or down contributing angular momentum and, at least relative to other spins in the atom, fine energy level.  Also allows 2 electrons per orbital without breaking the Pauli exclusion principle.
