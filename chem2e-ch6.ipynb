### Chapter 6 &ndash; Formulas and Concepts

$$ \begin{align}
\lambda &:= \text{wavelength} \\
\nu &:= \text{frequency (Greek letter nu)} \\
a &:= \text{amplitude (one half from trough to peak)} \\
\text{speed} &= \lambda \nu \\
c &= 2.998 \times 10^8 ms^{-1} \\
    &= \lambda\nu \text{ (for light)} \\
\\
J &= 1\ kg \frac{m^2}{s^2} \text{( note the kilograms)} \\
h &= 6.626 \times 10^{-34}\ J\ s \text{ (Planck's constant)} \\
E &:= \text{energy} \\
E &= h\nu \text{ (for one photon)} \\
    &= \frac{hc}{\lambda} \text{ (for one photon)} \\
\\
R_\infty &= \frac{k}{hc} = 1.097 \times 10^7\ m^{-1} \text{ (Rydberg constant)} \\
\frac{1}{\lambda} &= R_\infty \left(\frac{1}{n_1^2} - \frac{1}{n_2^2}\right)
    ,\ n_1 < n_2,\ n_1 \&\ n_2 \text{ integers (Rydberg equation)} \\
Z &= \text{atomic number (1 for H, 2 for He, etc.)} \\
k &= 2.179 \times 10^{-18}\ J \\
a_0 &= 5.292 \times 10^{-11}\ m \text{ (Bohr radius)} \\
\left|\Delta E\right| &= \left|E_f - E_i\right| = h\nu = \frac{hc}{\lambda} \\
\Delta E &= k \left( \frac{1}{n_1^2} - \frac{1}{n_2^2} \right) = \frac{hc}{\lambda} \\
E_n &= - \frac{kZ^2}{n^2} \text{ (orbital energies for atoms/ions with 1 electron)} \\
r &= \frac{n^2}{Z} a_0 \text{ (radius for same)} \\
E &= 0 \text{ is the ionization limit where the electron is freed} \\
\Delta E &= E_{n\to\infty} - E_1 = 0 + k = k \text{ (ionization energy from n=1)} \\
\\
\nu &:= \text{frequency} \\
v &:= \text{velocity} \\
p &:= \text{momentum} \\
\lambda &= \frac{h}{p} = \frac{h}{mv} \text{ (de Broglie wavelength for particles)} \\
\end{align} $$

#### Example 6.5 - Solution
What is the energy (in joules) and the wavelength (in meters) of the line in the spectrum of hydrogen that represents the movement of an electron from Bohr orbit with n = 4 to the orbit with n = 6? In what part of the electromagnetic spectrum do we find this radiation?

```python
E_4 = - k / 4**2
E_6 = - k / 6**2
E_4to6 = E_6 - E_4
E_4to6_b = float(k) * ( 1/4**2 - 1/6**2)
l = h * sc / E_4to6
display(E_4, E_6, E_4to6, E_4to6_b, l)
```
