**Main group elements** aka **representative elements** - 
Groups 1, 2, and 13 through 18, where last electron added is in an **s** or **p** orbital.  Valance electrons are those in highest shell ***n***.  (This means that **d** electrons added in the same periodic table row are core, not valence, electrons, because their shell is *n - 1*.)  Cations formed by removing last-added electrons.

**Transition elements** aka **transition elements** - 
Typically groups 3 through 12 (though IUPAC makes exceptions), where last electron added is in a **d** orbital.  Valence electrons are those in ***n*s** and **(*n - 1*)d**.  Cations formed by removing ***n*s** electrons first.

**Inner transition elements** - 
Lanthanide and Actinide series, where last electron added is in an **f** orbital, plus Lanthanum and Actinium themselves.  Valence electrons are those in ***n*s**, **(*n - 1*)d**, and **(*n - 2*)f**.  Cations formed by removing ***n*s** electrons first and then [maybe, unclear to me but example 6.11(e) points that way] the **(*n - 1*)d** electrons.

#### Example 6.5 - Solution
What is the energy (in joules) and the wavelength (in meters) of the line in the spectrum of hydrogen that represents the movement of an electron from Bohr orbit with n = 4 to the orbit with n = 6? In what part of the electromagnetic spectrum do we find this radiation?

```python
E_4 = - k / 4**2
E_6 = - k / 6**2
E_4to6 = E_6 - E_4
E_4to6_b = float(k) * ( 1/4**2 - 1/6**2)
l = h * sc / E_4to6
display(E_4, E_6, E_4to6, E_4to6_b, l)
```
