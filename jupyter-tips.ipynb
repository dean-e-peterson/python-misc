## Displaying Variables in Markdown or LaTeX

Unfortunately it is not as simple as in R Studio.  An extension that did it, Python Markdown, may not work with Jupyter Lab and the latest versions of Jupyter Notebook?  However, try this.

````python
from IPython.display import Markdown as md
var = 12
display(md(f'My variable `var` is {var}.  Or, at the end of a cell, just...'))
md(f'My variable `var` is {var}.')
````

LaTeX can be done with the same method.

````python
display(md('My variable $var=' + str(var) + '$' + f' or my variable $var={var}$'))
display(md('But combining f"" string and LaTeX curly braces gets trickier.'))
display(md(rf'$$ \frac{{24}}{{{var}}} = 2$$'))
display(md('Format strings also work, as $var={}$ or $var={v}$'.format(var, v=var)))
````

Or LaTeX can be done with `Math` instead of `Markdown`, but I'm not sure what it buys you.

````python
from IPython.display import display, Math
display(Math(rf'var={var} \text{{ or }} \frac{{24}}{{{var}}} = 2'))
display(md('or at the end of a cell, just...'))
Math(rf'var={var}')
````

$\text{How to put a subscript under a degree sign:}$

$$ \begin{align}
\text{Degree sign with \\degree:} \quad &25\ \degree C \\
\text{Degree sign with superscripted \\circ:} \quad &25\ ^\circ C \\
\text{Degree \\degree with subscript:} \quad &\Delta H \degree _f \\
\text{Degree \\circ superscripted with subscript:} \quad &\Delta H^\circ_f \\
\text{Degree \\degree \\rlap with subscript:} \quad &\Delta H \rlap{\degree}_f \\
\text{The same but one fewer set of braces:} \quad &\Delta H \rlap \degree _f \\
\text{\\underset did not seem to work well for this.} \quad &\\
\end{align} $$

Note that with `\rlap` in real TeX environments you may have to go back into math mode with $ around the first(?) argument? You don't have to in Jupyter Notebooks because of [MathJax differences](https://docs.mathjax.org/en/latest/input/tex/differences.html).

````python
%%javascript
// Dynamically load THREE module to avoid import-map issues.
// Probably depends on Jupyter already having threejs import-mapss.
import('three').then(THREE => {
import('three/addons/controls/OrbitControls.js').then(ORBIT => {
    loaded(THREE, ORBIT);
});
});

function loaded(THREE, ORBIT)
{
    const canvas = setUpCanvas();

    const scene = new THREE.Scene();
    scene.background = new THREE.Color('#fafafa');

    const fov = 35; // Field of view
    const aspect = canvas.width / canvas.height;
    const near = 0.1; // Near clipping plane
    const far = 100; // Far clipping plane
    const camera = new THREE.PerspectiveCamera(fov, aspect, near, far);
    //camera.updateProjectionMatrix(); // update the camera's frustum, for resize.
    //const camera = new THREE.OrthographicCamera(-3, 3, 2, -2, 0.1, 20);
    camera.position.set(3, 3, 3);
    camera.lookAt(0, 0, 0);

    const intensity = 5;
    const light = new THREE.DirectionalLight('white', intensity);
    light.position.set(-10, 10, 10); // And points at 0, 0, 0
    //scene.add(light);
    // Camera does not need to be added to a scene, but CAN be so you can
    // make things like lights children of the camera to move with the camera.
    scene.add(camera);
    camera.add(light);

    const axesSize = 3;
    const axes = new THREE.AxesHelper(axesSize);
    scene.add(axes);

    const gridSize = 4;
    const gridDivisions = 8;
    const grid = new THREE.GridHelper(gridSize, gridDivisions);
    scene.add(grid);

    const material = new THREE.MeshStandardMaterial({
        color: new THREE.Color('slategray'),
        transparent: true,
        opacity: 0.5,
        // side: THREE.DoubleSide,
        wireframe: false, 
    });
    const nuclearMaterial = new THREE.MeshStandardMaterial({
        color: new THREE.Color('black'),
        // transparent: true,
        // opacity: 0.5,
        // side: THREE.DoubleSide,
        wireframe: false, 
    });
    const sMaterial = new THREE.MeshStandardMaterial({
        color: new THREE.Color('deepskyblue'),
        transparent: true,
        opacity: 0.5,
        // side: THREE.DoubleSide,
        wireframe: false, 
    });
    const pMaterialPhaseUp = new THREE.MeshStandardMaterial({
        color: new THREE.Color('DarkTurquoise'),
        transparent: true,
        opacity: 0.5,
        // side: THREE.DoubleSide,
        wireframe: false,
    });
    const pMaterialPhaseDown = new THREE.MeshStandardMaterial({
        color: new THREE.Color('lightcoral'),
        transparent: true,
        opacity: 0.5,
        // side: THREE.DoubleSide,
        wireframe: false,
    });

    const oxygen = new THREE.Group();

    const nuclearRadius = .05;
    const nuclearGeometry = new THREE.SphereGeometry(nuclearRadius);
    const nuclearSphere = new THREE.Mesh(nuclearGeometry, nuclearMaterial);
    oxygen.add(nuclearSphere);

    const sRadius = .25;
    const sGeometry = new THREE.SphereGeometry(sRadius);
    const sSphere = new THREE.Mesh(sGeometry, sMaterial);
    oxygen.add(sSphere);

    const p = new THREE.Group();
    const pTipDistance = .5;
    const pSphereRadius = .2;
    const pConeTipAngle = Math.asin(pSphereRadius / pTipDistance);
    const pConeBottomRadius = pSphereRadius * Math.cos(pConeTipAngle);
    const pConeHeight = pConeBottomRadius / Math.tan(pConeTipAngle);
    const pConeTopRadius = 0;

    const pWidthSegments = 32;
    const pHeightSegments = 16;
    const pPhiStart = 0;
    const pPhiLength = Math.PI * 2; // Math.PI * 2;
    const pThetaStart = 0;
    const pThetaLength = (Math.PI / 2) + pConeTipAngle; // Math.PI;
    const pSphereGeometry = new THREE.SphereGeometry(
        pSphereRadius,
        pWidthSegments,
        pHeightSegments,
        pPhiStart,
        pPhiLength,
        pThetaStart,
        pThetaLength
    );
    const pSpherePhaseUp = new THREE.Mesh(pSphereGeometry, pMaterialPhaseUp);
    const pSpherePhaseDown = new THREE.Mesh(pSphereGeometry, pMaterialPhaseDown);
    pSpherePhaseUp.position.x = pTipDistance;
    pSpherePhaseDown.position.x = - pTipDistance;
    pSpherePhaseUp.rotation.z = - Math.PI / 2;
    pSpherePhaseDown.rotation.z = Math.PI / 2;
    p.add(pSpherePhaseUp);
    p.add(pSpherePhaseDown);

    const pConeRadialSegments = 32;
    const pConeHeightSegments = 1;
    const pConeOpenEnded = true;
    const pConeGeometry = new THREE.CylinderGeometry(
        pConeTopRadius,
        pConeBottomRadius,
        pConeHeight,
        pConeRadialSegments,
        pConeHeightSegments,
        pConeOpenEnded
    );
    const pConePhaseUp = new THREE.Mesh(pConeGeometry, pMaterialPhaseUp);
    const pConePhaseDown = new THREE.Mesh(pConeGeometry, pMaterialPhaseDown);
    pConePhaseUp.position.x = pConeHeight / 2;
    pConePhaseDown.position.x = - pConeHeight / 2;
    pConePhaseUp.rotation.z = Math.PI / 2;
    pConePhaseDown.rotation.z = - Math.PI / 2;
    p.add(pConePhaseUp);
    p.add(pConePhaseDown);
    
    const px = p.clone();    
    const py = p.clone();
    py.rotation.z = Math.PI / 2;
    const pz = p.clone();
    pz.rotation.y = - Math.PI / 2;
    oxygen.add(px, py, pz);

    const o1 = oxygen.clone();
    o1.position.x = 1;
    const o2 = oxygen.clone();
    o2.position.x = -1;
    o2.scale.x = -1;
    scene.add(o1, o2);

    const renderer = new THREE.WebGLRenderer({
        canvas: canvas, // Since not adding renderer.domElement to container.
        antialias: true,
    });
    renderer.physicallyCorrectLights = true; // Requires 1 unit = 1 meter
    renderer.setSize(canvas.width, canvas.height);
    renderer.setPixelRatio(window.devicePixelRatio);

    /* !!! Don't render to save resources when refreshing whole notebook.
    const controls = new ORBIT.OrbitControls(camera, canvas);
    controls.addEventListener('change', event => {
        renderer.render(scene, camera);
    });

    // container.append(renderer.domElement); // Add <canvas> to container
    renderer.render(scene, camera);
    */
}

function setUpCanvas()
{
    const canvas = document.createElement('canvas');
    canvas.width = 600;
    canvas.height = 400;
    canvas.style = 'background: #fafafa; width: 600; height: 400;';
    element.append(canvas);
    return canvas;
}
````
