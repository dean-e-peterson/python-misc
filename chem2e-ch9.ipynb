### Example 9.1 - Solution
The United States National Weather Service reports pressure in both inches of Hg and millibars. Convert a pressure of 29.2 in. Hg into:

(a) torr

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ atm}{101,325\ Pa} \cdot
    \frac{720\ torr}{1\ atm}
    = 742\ torr
$$

(b) atm

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ atm}{101,325\ Pa}
    = 0.976\ atm
$$

(c) kPa

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ kPa}{1000\ Pa}
    = 98.9\ kPa
$$

(d) mbar

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ bar}{100,000\ Pa} \cdot
    \frac{1000\ mbar}{1\ bar}
    = 989\ mbar
$$


```python
display(29.2 * 3386 / 101325 * 760)
display(29.2 * 3386 / 101325)
display(29.2 * 3386 / 1000)
display(29.2 * 3386 / 100000 * 1000)
display('---')
display(29.2 * 25.4)
display(742 * 101325 / 760)
```

$$ 740\ torr \approx 740\ mm.\ Hg $$

bar

$$ 740\ torr \cdot
