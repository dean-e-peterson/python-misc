## Chapter 9 &ndash; Units

\begin{align}
N &= 1\ \frac{kg\ m}{s^2} := \text{Newton} \\
Pa &= 1\ \frac{N}{m^2} := \text{Pascal} \\[1ex]
psi &\approx 6.895\ kPa := \text{pounds per square inch} \\[1.5ex]
atm &= 101,325\ Pa = 760\ torr \approx 14.7\ psi := \text{atmosphere} \\[1.5ex]
bar\ \text{or}\ b &= 100,000\ Pa \text{ exactly} \\
torr &= \frac{1}{760}\ atm \\[1ex]
mm\ Hg &\approx 1\ torr \\[1.5ex]
in.\ Hg &= 3386\ Pa\\[1.5ex]
\end{align}

## Chapter 9 &ndash; Equations

\begin{align}
P &= \frac{F}{A} := \text{Pressure} \\
h &:= \text{height of fluid column} \\
\rho &:= \text{fluid density} \\
g &= 9.81\ m/s^2:= \text{acceleration due to gravity} \\
p &= h \rho g := \text{hydrostatic pressure} \\[1ex]
&\textbf{Amonton or Guy-Lussac's Law} \text{ - Constant volume} \\[1ex]
P &\propto T_\text{Kelvin} \\
P &= \text{constant} \times T_\text{Kelvin} \\
P &= k \times T_\text{Kelvin} \\
k &= \frac{P}{T}, \quad \frac{P_1}{T_1} = \frac{P_2}{T_2} \\[1ex]
&\textbf{Charles's Law} \text{ - Constant pressure} \\[1ex]
V &\propto T_\text{Kelvin} \\
V &= k \times T_\text{Kelvin} \\
k &= \frac{V}{T}, \quad \frac{V_1}{T_1} = \frac{V_2}{T_2} \\[1ex]
\end{align}

### Example 9.1 - Solution
The United States National Weather Service reports pressure in both inches of Hg and millibars. Convert a pressure of 29.2 in. Hg into:

(a) torr

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ atm}{101,325\ Pa} \cdot
    \frac{720\ torr}{1\ atm}
    = 742\ torr
$$

(b) atm

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ atm}{101,325\ Pa}
    = 0.976\ atm
$$

(c) kPa

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ kPa}{1000\ Pa}
    = 98.9\ kPa
$$

(d) mbar

$$\quad 29.2\ in.\ Hg \cdot 
    \frac{3386\ Pa}{1\ in.\ Hg} \cdot
    \frac{1\ bar}{100,000\ Pa} \cdot
    \frac{1000\ mbar}{1\ bar}
    = 989\ mbar
$$


```python
display(29.2 * 3386 / 101325 * 760)
display(29.2 * 3386 / 101325)
display(29.2 * 3386 / 1000)
display(29.2 * 3386 / 100000 * 1000)
display('---')
display(29.2 * 25.4)
display(742 * 101325 / 760)
```

### Example 9.2 - Solution
Show the calculation supporting the claim that atmospheric pressure near sea level corresponds to the pressure exerted by a column of mercury that is about 760 mm high. The density of mercury = 13.6 g/cm<sup>3</sup>.

\begin{align}
N &= 1\ \frac{kg\ m}{s^2} := \text{Newton} \\[1ex]
Pa &= 1\ \frac{N}{m^2} := \text{Pascal} \\[1ex]
\rho_{[\ce{Hg}]} &= \left(13.6 \frac{g}{cm^3}\right)
        \left(\frac{1\ kg}{1000\ g}\right)
        \left(\frac{(100\ cm)^3}{m^3}\right) 
    = 13,600 \frac{kg}{m^3} \\[1ex]
p &= h \rho g \\[1ex]
h &= \frac{p}{\rho g} \\
  &= \frac{101,325\ Pa}{13.6\ g/cm^3 \cdot 9.81\ m/s^2} \\
  &= 101,325\
      \left(\frac{kg\ m}{s^2}\right)
      \left(\frac{1}{m^2}\right)
      \left(\frac{1\ m^3}{13,600\ kg}\right)
      \left(\frac{1\ s^2}{9.81\ m}\right)
      \left(\frac{1000\ mm}{1\ m}\right) \\
  &= 759\ mm \\
\end{align}

```python
display(101325 / 13600 / 9.81 * 1000)
display(101325 / 13600 / 9.80665 * 1000)
display(101325 / 13546 / 9.80665 * 1000)
display(101325 / 13628.22 / 9.80665 * 1000)
display('book worked the problem the other way, from mm to Pa')
display(.760 * 13600 * 9.81)
display(.760 * 13600 * 9.80665)
display(.760 * 13546 * 9.80665)
display(.760 * 13628.22 * 9.80665)
```

### Example 9.5 - Solution

**Predicting Change in Pressure with Temperature**

A can of hair spray is used until it is empty except for the propellant, isobutane gas.

(a) On the can is the warning “Store only at temperatures below 120 °F (48.8 °C). Do not incinerate.” Why?

(b) The gas in the can is initially at 24 °C and 360 kPa, and the can has a volume of 350 mL. If the can is left in a car that reaches 50 °C on a hot day, what is the new pressure in the can?

```python
display(24 + 273)
display(50 + 273)
display(360 * (50 + 273) / (24 + 273))
display(360 * (50 + 273.15) / (24 + 273.15))
display(360 * 323 / 297)
display(360 * 323)
display(116000 / 297)
display(323 / 297)
display(360 * 1.09)
```

$$ \frac{24 + 273 K}{360\ kPA} = \frac{50 + 273 K}{P_{new}} $$
$$ P_{new} = 360\ kPa \times \frac{50 + 273 K}{24 + 273 K} = 392\ kPa $$

(a) In addition to isobutane being flammable, the isobutante is stored in a constant volume, so if the temperature gets too high, the pressure can get too high and burst the can.

(b) 392 kPa
