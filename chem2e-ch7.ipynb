## Chapter 7 &ndash; Formulas and Concepts

\begin{align}

**cation** - Positive ion.

**anion** - Negative ion.

**ionic compound** - aka **salt**
- Compound composed of ions
- Electrons transferred
- High electronegativity difference (often > 1.8)
- Not molecular
- **Common properties**
  - When solid
    - crystalline structure
    - rigid and brittle
    - ionic bonds are strong, so...
    - high melting and boiling points
    - poor electrical conductors - "ions" [electrons?] can't move
    - dissolve in water
  - Dissolved or melted
    - good electrical and heat conductors - ions move freely

- **Cation charge rules**
    - Groups 1 and 2, same as group number
    - Groups 13 through 17, group number minus 10, but also
    - s valence electrons are first to go
      - **inert pair effect** - Heavy ion s valence electron low energy can cause positive ions with two fewer positive charges than expected toward the bottom of groups 13, 14, and 15...and $\ce{Hg_2^{2+}}$.
      - Transition metals 2+ or 3+ from losing valence s electrons and possibly a d electron.
      - Inner transition metals 3+ from losing valence s electrons and a d or f electron.

- **Anion charge rules**
    - Take a non-metal, fill the s and p orbitals to get the same electron configuration as the next noble gas.

**Covalent bond**
- Electrons shared.
- Atoms of similar attraction for electrons.
- Polyatomic ions like $\ce{NH^4+}$ and $\ce{NO^3-}$ are held together covalently within themselves.
- **Common properties**
    - attraction weaker than ionic attraction, so...
    - lower melting and boiling points
    - softer when solid
    - poor electrical conductors in any state
    - do not dissolve in water

- **Pure covalent bond**
    - Electrons shared equally
    - Low electronegativity difference (often < 0.4)
    - Example: a bond between atoms of the same element

- **Polar covalent bond**
    - Electron (probability distributions) not shared equally, resulting in...
    - Intermediate electronegativity difference (often 0.4 - 1.8)
    - Partial positive/negative charges on the atoms

**electronegativity** - Measure of an atom's tendency to attract electrons.
- Increases left to right
- Decreases down a group
- Nobel gasses, group 18, excluded
- Dimensionless
- *arbitrary* or at least calculated, not measured, scale


**lone pairs** - Pairs of electrons not used in bonding.

**Lewis symbol** - A element symbol surrounded by one dot for each valence electron.
Like $ :\!\overset{\centerdot\centerdot}{\underset{\centerdot}{\ce{S}}}\,\cdot $

**Lewis structure** - Diagrams showing bonding using Lewis symbols, like
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$
or
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}-
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$

**single bond** - Single pair of covalently shared electrons, like
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$
or
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}-
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$

**double bond** - Maybe like (guessing structure, recording notation)
$
:\!\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}::
\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}:
$
or
$
:\!\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}=
\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}:
$

**triple bond** - Like $ :\ce{C}:::\ce{O}: $
or $ :\ce{C} \equiv \ce{O}: $

**octet rule** - "tendency of main group atoms to form enough bonds to obtain 8 valence electrons"

**octet rule exceptions**
- **free radicals** - Molecules with an odd number of valence electrons.
- Molecules that are electron-deficient, with a central atom having fewer electrons than a noble gas.  The central atom is usually group 2 or 13.  The outer atoms may be hydrogen.  Tend to be very reactive molecules.
- Molecules that are hypervalent, with a central atom that has more electrons than a noble gas. Central atom from period >=3 using empty d orbitals in same shell.


**Lewis diagram steps from book**
1. Determine total number of valence electrons.
2. Place an atom with the lowest electronegativity in the center, and add single bonds to the other atoms.
3. Distribute remaining valence electrons as lone pairs to terminal atoms other than hydrogen.
4. Place remaining valence electrons around central atom.
5. Rearrange electrons of outer atoms to make double or triple bonds with the central atom to obtain octets.
6. Add any remaining electrons to the central atom (from section on hypervalent molecules.)

**formal charge** - "The formal charge of an atom in a molecule is the hypothetical charge the atom would have if we could redistribute the electrons in the bonds evenly between the atoms."  Not the actual charge, just a tool.  Sum to charge of molecule (0) or ion.

\begin{align}
\text{formal charge} &= \text{\# of valence electrons in free atom} \\
    &- \text{\# of lone pair electrons} \\
    &- 1/2 \text{ \# of bonding electrons (which equals \# of bonds)}
\end{align}

**Choosing preferred Lewis diagram based on formal charges quoted from book**
1. "A molecular structure in which all formal charges are zero is preferable"
2. "the arrangement with the smallest nonzero formal charges is preferable."
3. "preferable when adjacent formal charges are zero or of the opposite sign."
4. "the structure with the negative formal charges on the more electronegative atoms is preferable."

**resonance** - "If two or more Lewis structures with the same arrangement of atoms can be written for a molecule or ion, the actual distribution of electrons is an average of that shown by the various Lewis structures."

**resonance forms** - The individual Lewis structures being hybridized.  Not real.  (Dragon and unicorn.)

**resonance hybrid** - The electronic structure averaging the forms.  Real.  (Rhinoceros.)

C–H, C–N, C–O, N–H, O–H, S–H

&\Delta EN(\ce{C-H}) &&= -0.4 \\
&\Delta EN(\ce{C-N}) &&= 0.5 \\

\end{alignat}

\Delta EN(\ce{\overset{\delta -}{C} - \overset{\delta +}{H}})
= \Delta EN(\ce{\overset{\delta -}{S} - \overset{\delta +}{H}})
< \Delta EN(\ce{\overset{\delta +}{C} - \overset{\delta -}{N}})
< \Delta EN(\ce{\overset{\delta -}{N} - \overset{\delta +}{H}})
< \Delta EN(\ce{\overset{\delta +}{C} - \overset{\delta -}{O}})
< \Delta EN(\ce{\overset{\delta -}{O} - \overset{\delta +}{H}})
$$
