**cation** - Positive ion.

**anion** - Negative ion.

**lone pairs** - Pairs of electrons not used in bonding.

**octet rule** - "tendency of main group atoms to form enough bonds to obtain 8 valence electrons"

**Lewis symbol** - A element symbol surrounded by one dot for each valence electron.
Like $ :\!\overset{\centerdot\centerdot}{\underset{\centerdot}{\ce{S}}}\,\cdot $

**Lewis structure** - Diagrams showing bonding using Lewis symbols, like
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$
or
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}-
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$

**single bond** - Single pair of covalently shared electrons, like
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$
or
$
:\!\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}-
\overset{\centerdot\centerdot}{\underset{\centerdot\centerdot}{\ce{Cl}}}:
$

**double bond** - Maybe like (guessing structure, recording notation)
$
:\!\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}::
\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}:
$
or
$
:\!\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}=
\overset{\centerdot}{\underset{\centerdot}{\ce{O}}}:
$

**triple bond** - Like $ :\ce{C}:::\ce{O}: $
or $ :\ce{C} \equiv \ce{O}: $


**ionic compound** - aka **salt**
- Compound composed of ions
- Electrons transferred
- High electronegativity difference (often > 1.8)
- Not molecular
- **Common properties**
  - When solid
    - crystalline structure
    - rigid and brittle
    - ionic bonds are strong, so...
    - high melting and boiling points
    - poor electrical conductors - "ions" [electrons?] can't move
    - dissolve in water
  - Dissolved or melted
    - good electrical and heat conductors - ions move freely

- **Cation charge rules**
    - Groups 1 and 2, same as group number
    - Groups 13 through 17, group number minus 10, but also
    - s valence electrons are first to go
      - **inert pair effect** - Heavy ion s valence electron low energy can cause positive ions with two fewer positive charges than expected toward the bottom of groups 13, 14, and 15...and $\ce{Hg_2^{2+}}$.
      - Transition metals 2+ or 3+ from losing valence s electrons and possibly a d electron.
      - Inner transition metals 3+ from losing valence s electrons and a d or f electron.

- **Anion charge rules**
    - Take a non-metal, fill the s and p orbitals to get the same electron configuration as the next noble gas.

**Covalent bond**
- Electrons shared.
- Atoms of similar attraction for electrons.
- Polyatomic ions like $\ce{NH^4+}$ and $\ce{NO^3-}$ are held together covalently within themselves.
- **Common properties**
    - attraction weaker than ionic attraction, so...
    - lower melting and boiling points
    - softer when solid
    - poor electrical conductors in any state
    - do not dissolve in water

- **Pure covalent bond**
    - Electrons shared equally
    - Low electronegativity difference (often < 0.4)
    - Example: a bond between atoms of the same element

- **Polar covalent bond**
    - Electron (probability distributions) not shared equally, resulting in...
    - Intermediate electronegativity difference (often 0.4 - 1.8)
    - Partial positive/negative charges on the atoms

**electronegativity** - Measure of an atom's tendency to attract electrons.
- Increases left to right
- Decreases down a group
- Nobel gasses, group 18, excluded
- Dimensionless
- *arbitrary* or at least calculated, not measured, scale

### Example 7.3 - Solution
Bond polarities play an important role in determining the structure of proteins. Using the electronegativity values in Figure 7.6, arrange the following covalent bonds—all commonly found in amino acids—in order of increasing polarity. Then designate the positive and negative atoms using the symbols δ+ and δ–:
C–H, C–N, C–O, N–H, O–H, S–H

\begin{alignat}{2}
&\Delta EN(\ce{C-H}) &&= -0.4 \\
&\Delta EN(\ce{C-N}) &&= 0.5 \\
&\Delta EN(\ce{C-O}) &&= 1 \\
&\Delta EN(\ce{N-H}) &&= -0.9 \\
&\Delta EN(\ce{O-H}) &&= -1.4 \\
&\Delta EN(\ce{S-H}) &&= -0.4 \\
\end{alignat}
$$
\Delta EN(\ce{\overset{\delta -}{C} - \overset{\delta +}{H}})
= \Delta EN(\ce{\overset{\delta -}{S} - \overset{\delta +}{H}})
< \Delta EN(\ce{\overset{\delta +}{C} - \overset{\delta -}{N}})
< \Delta EN(\ce{\overset{\delta -}{N} - \overset{\delta +}{H}})
< \Delta EN(\ce{\overset{\delta +}{C} - \overset{\delta -}{O}})
< \Delta EN(\ce{\overset{\delta -}{O} - \overset{\delta +}{H}})
$$

### Example 7.3 - Check Your Learning
Silicones are polymeric compounds containing, among others, the following types of covalent bonds: Si–O, Si–C, C–H, and C–C. Using the electronegativity values in Figure 7.6, arrange the bonds in order of increasing polarity and designate the positive and negative atoms using the symbols δ+ and δ–.

\begin{alignat}{2}
&\Delta EN(\ce{Si-O}) &&= -1.7 \\
&\Delta EN(\ce{Si-C}) &&= -0.7 \\
&\Delta EN(\ce{C-H}) &&= 0.4 \\
&\Delta EN(\ce{C-C}) &&= 0 \\
\end{alignat}
$$
\Delta EN(\ce{\overset{\text{nonpolar}}{C - C}})
< \Delta EN(\ce{\overset{\delta -}{C} - \overset{\delta +}{H}})
< \Delta EN(\ce{\overset{\delta +}{Si} - \overset{\delta -}{C}})
< \Delta EN(\ce{\overset{\delta +}{Si} - \overset{\delta -}{O}})
$$
