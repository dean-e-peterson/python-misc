### Chapter 8 &ndash; Notes

**VSEPR theory** - **Valence Shell Electron-Pair Repulsion theory** - Approximately predicts molecular structure shape by considering repulsion between valence electron pairs, both bonded pairs and lone pairs.

---

**Valence bond theory** - A covalent bond is where:
1. "an orbital on one atom overlaps an orbital on a second atom"
2. "the single electrons in each orbital combine to form an electron pair"
- "mutual attraction between this negatively charged electron pair and the two atoms’ positively charged nuclei serves to physically link the two atoms"
- "strength of a covalent bond depends on the extent of overlap"
- "the overlap is greater when the orbitals overlap end to end rather than at an angle"
- without hybridization, only accurate for two atom molecules?

**sigma bond** - **$\mathbf{\large{\sigma}}$ bond** - Electron density along the axis between nuclei.  From overlaps of orbitals:
- s and s
- s and p
- p and p (inline)
- hybridized orbitals (see below)

**pi bond** - **$\mathbf{\large{\pi}}$ bond** - Electron densities to the sides of axis between nuclei.  Along axis between nuclei is a planar **node** of zero electron probability.  Generally less overlap and less bond strength than $\sigma$ bonds. From overlaps of orbitals:
- p and p (side by side)

Single bond - 1 $\sigma$ bond

Double bond - 1 $\sigma$ and 1 $\pi$ bond

Triple bond - 1 $\sigma$ and 2 $\pi$ bonds

**hybrid orbital** - Orbital in a molecule different from that of a lone component atom.  The book says...

> The following ideas are important in understanding hybridization:
> 1. Hybrid orbitals do not exist in isolated atoms. They are formed only in covalently bonded atoms.
> 2. Hybrid orbitals have shapes and orientations that are very different from those of the atomic orbitals in isolated atoms.
> 3. A set of hybrid orbitals is generated by combining atomic orbitals. The number of hybrid orbitals in a set is equal to the number of atomic orbitals that were combined to produce the set.
> 4. All orbitals in a set of hybrid orbitals are equivalent in shape and energy.
> 5. The type of hybrid orbitals formed in a bonded atom depends on its electron-pair geometry as predicted by the VSEPR theory.
> 6. Hybrid orbitals overlap to form σ bonds. Unhybridized orbitals overlap to form π bonds.

$\large{sp}$ - Hybrid of one s orbital and one p orbital.  Two $sp$ orbitals are arranged in a linear geometry. "Involves mixing of the valence s orbital with one of the valence p orbitals".  Results in a two orbitals with intermediate energy level in a linear configuration.  "Any central atom surrounded by just two regions of valence electron density in a molecule will exhibit sp hybridization."

$\large{sp^2}$ - Hybrid of one s orbital and two p orbitals. Three $sp^2$ orbitals are arranged in a 120&deg; trigonal planar geometry.

$\large{sp^3}$ - Hybrid of one s orbital and three p orbitals. Four $sp^3$ orbitals are arranged in a tetrahedral geometry.

\begin{align}
\mathrm{count}_\sigma &= \mathrm{count}_{\text{all bonds}} \text{(double and triple only count for 1)} \\
\mathrm{count}_\pi &= \mathrm{count}_{\text{non-single bonds}}
\end{align}
